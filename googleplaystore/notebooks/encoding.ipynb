```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns 
```

```python
df=pd.read_csv('../data/googleplaystore.csv')
df.head()
```

```python
clean_df=df.copy()
clean_df.head()
```

```python
clean_df["Reviews"].str.isnumeric().value_counts()
```

```python
clean_df[~clean_df["Reviews"].str.isnumeric()]
```

```python
clean_df=clean_df.drop(clean_df.index[10472])
```

```python
clean_df[~clean_df["Reviews"].str.isnumeric()]
```

```python
clean_df.info()
```

```python
clean_df["Size"]=clean_df["Size"].str.replace("M", "000")
clean_df["Size"]=clean_df["Size"].str.replace("k", "")
```

```python
clean_df["Size"]=clean_df["Size"].replace(
    'Varies with device', np.nan)
```

```python
clean_df["Size"].unique()
```

```python
clean_df["Size"].value_counts()
```

```python
# "Price" and "Installs" columns -> remove "+", "$" and ","

remove=["+",",","$"]
columns=["Installs","Price"]
for item in remove:
    for column in columns:
        clean_df[column]=clean_df[column].str.replace(item,"")
```

```python
clean_df["Price"].unique()
```

```python
clean_df["Installs"].unique()
```

```python
clean_df["Size"]=clean_df["Size"].astype(float)
clean_df["Price"]=clean_df["Price"].astype(float)
clean_df["Installs"]=clean_df["Installs"].astype(int)
```

```python
clean_df.info()
```

```python
clean_df["Last Updated"]=pd.to_datetime(clean_df["Last Updated"])
clean_df["Day"]=clean_df["Last Updated"].dt.day
clean_df["Month"]=clean_df["Last Updated"].dt.month
clean_df["Year"]=clean_df["Last Updated"].dt.year
```
